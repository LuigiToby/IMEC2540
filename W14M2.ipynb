```python
x = dynamicsymbols("x")

m,c,k,t = symbols("m c k t")

zeta, omega_n = symbols("zeta omega_n")

eq1 = m*x.diff(t,t)/m + c*x.diff(t)/m + k*x/m

eq2 = x.diff(t,t) + 2*zeta*omega_n*x.diff(t) + omega_n**2*x

display(eq1, eq2)

eq3  = eq2.subs({x.diff(t,t):0, x.diff(t):1, x:0})

eq4 = eq1.subs({x.diff(t,t):0, x.diff(t):1, x:0})


eq5  = eq2.subs({x.diff(t,t):0, x.diff(t):0, x:1})

eq6 = eq1.subs({x.diff(t,t):0, x.diff(t):0, x:1})

coef1 = eq3-eq4
coef2 = eq5-eq6


display([coef1, coef2])
```
